```python
import pymupdf
import matplotlib.pyplot as plt
import matplotlib.patches as patches

def visualize_pdf(pdf_path, page_number=0):
    doc = pymupdf.open(pdf_path)
    page = doc[page_number]

    # Extract text information
    data = page.get_text("dict")

    fig, ax = plt.subplots(figsize=(12, 16))

    # Display original PDF page
    pix = page.get_pixmap(dpi=120, alpha=False)

    import io
    from PIL import Image

    img = Image.open(io.BytesIO(pix.tobytes("png")))

    ax.imshow(img)

    # Scale factor: rendered image / PDF page dimensions
    scale_x = pix.width / page.rect.width
    scale_y = pix.height / page.rect.height

    for b, block in enumerate(data["blocks"]):
        x0, y0, x1, y1 = block["bbox"]
        # Convert PDF coordinates to image coordinates
        x0 *= scale_x
        y0 *= scale_y
        x1 *= scale_x
        y1 *= scale_y
        rect = patches.Rectangle(
            (x0, y0),
            x1 - x0,
            y1 - y0,
            linewidth=1,
            edgecolor="red",
            facecolor="none"
        )

        ax.add_patch(rect)
        ax.text(
            x0, y0 - 3,
            b,
            color="blue",
            fontsize=6
        )
        if "lines" not in block:
            continue

        for line in block["lines"]:
            for span in line["spans"]:

                x0, y0, x1, y1 = span["bbox"]
                print(span["size"], span["font"], end=": ")
                print(span["text"])
                fontname = span["font"].replace(",", " ")
                if "Aptos" in fontname:


                # Convert PDF coordinates to image coordinates
                x0 *= scale_x
                y0 *= scale_y
                x1 *= scale_x
                y1 *= scale_y

                rect = patches.Rectangle(
                    (x0, y0),
                    x1 - x0,
                    y1 - y0,
                    linewidth=1,
                    edgecolor="green",
                    facecolor="none"
                )

                ax.add_patch(rect)
        #
        #         # Display extracted text
        #         if "\uf0e0" in span["text"]:
        #             print(span["text"])
        #         ax.text(
        #             x0, y0 - 3,
        #             span["text"][:20],
        #             color="blue",
        #             fontsize=6
        #         )
        #
        #         print(span["size"], repr(span["text"]))

                # print(span["size"], span["flags"], span["bidi"], span["char_flags"], span["font"], span["color"], span["alpha"], span["ascender"], span["descender"], span["text"], span["origin"], span["bbox"])

    ax.set_xlim(0, pix.width)
    ax.set_ylim(pix.height, 0)
    ax.set_title(f"PDF Bounding Box Visualizer - Page {page_number + 1}")
    ax.axis("off")

    plt.tight_layout()
    plt.show()

    doc.close()

for i in range(27):
    visualize_pdf(SRC_PATH, page_number=i)
```
